```python
import matplotlib.pyplot as plt
import numpy as np

# Data
models = ['baseline', 'kham', 'lhasa']
cer = [0.1927, 0.2927, 0.2932]
ser = [0.4417, 0.5832, 0.5647]
wer = [0.3801, 0.5778, 0.5979]

# Create subplots
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))

# CER Histogram
ax1.bar(models, cer, color=['blue', 'orange', 'green'], alpha=0.7)
ax1.set_title('Character Error Rate (CER)')
ax1.set_ylabel('Error Rate')
ax1.set_ylim(0, 0.35)
ax1.grid(axis='y', alpha=0.3)

# Add value labels on bars
for i, v in enumerate(cer):
    ax1.text(i, v + 0.01, f'{v:.3f}', ha='center', va='bottom')

# SER Histogram
ax2.bar(models, ser, color=['blue', 'orange', 'green'], alpha=0.7)
ax2.set_title('Syllable Error Rate (SER)')
ax2.set_ylabel('Error Rate')
ax2.set_ylim(0, 0.65)
ax2.grid(axis='y', alpha=0.3)

for i, v in enumerate(ser):
    ax2.text(i, v + 0.01, f'{v:.3f}', ha='center', va='bottom')

# WER Histogram
ax3.bar(models, wer, color=['blue', 'orange', 'green'], alpha=0.7)
ax3.set_title('Word Error Rate (WER)')
ax3.set_ylabel('Error Rate')
ax3.set_ylim(0, 0.65)
ax3.grid(axis='y', alpha=0.3)

for i, v in enumerate(wer):
    ax3.text(i, v + 0.01, f'{v:.3f}', ha='center', va='bottom')

plt.tight_layout()
plt.show()

# Alternative: Combined histogram
plt.figure(figsize=(10, 6))
x = np.arange(len(models))
width = 0.25

plt.bar(x - width, cer, width, label='CER', alpha=0.7)
plt.bar(x, ser, width, label='SER', alpha=0.7)
plt.bar(x + width, wer, width, label='WER', alpha=0.7)

plt.xlabel('Models')
plt.ylabel('Error Rates')
plt.title('Error Rates Comparison Across Models')
plt.xticks(x, models)
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
